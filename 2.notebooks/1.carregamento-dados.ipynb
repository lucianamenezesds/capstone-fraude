```python
import pandas as pd
import numpy as np
```

### Carregamento de dados

```python
march = pd.read_excel(
    '../1.data/1.raw/compras_marco.xlsx'
)
```

```python
print(march.shape)
march.head()
```

```python
april = pd.read_csv("../1.data/1.raw/compras_abril.csv", sep=";", decimal=",")
print(april.shape)
april.head()
```

```python
march.info()
```

```python
april.info()
```

```python
documents = pd.read_json('../1.data/1.raw/documentos.json')
```

```python
print(documents.shape)
documents.head()
```

```python
documents.info()
```

```python
colunas = {
    "idCompra": "id_compra",
    "doc1": "entrega_doc_1",
    "doc2": "entrega_doc_2",
    "doc3": "entrega_doc_3"
}

documents = documents.rename(columns=colunas)
documents.head()
```

## Junção de dados

```python
## metodo concat() 

df_march_april = pd.concat([march, april], ignore_index=True)
print(df_march_april.shape)
df_march_april.head()
```

```python
## metodo merge() - unindo dataframes (similar ao JOIN do SQL)
```

```python
df_complete = pd.merge(df_march_april, documents, how="left", on="id_compra")
df_complete.info()
```

```python
df_complete[["entrega_doc_1","entrega_doc_2","entrega_doc_3"]].isna().sum()
```

```python
df_complete.to_csv("../1.data/2.processed/dados_marco_abril.csv", index=False)
```
