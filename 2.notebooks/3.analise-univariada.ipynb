## Histograma - numerica

-Distribuição de valores

```python
##bins - número de barras - 10 é o valor default

plt.hist(df_purchases['valor_compra'], edgecolor='black', bins=50, range=(0, 300))
plt.title("Histograma de distribuições de valores de compras (\\$0 a \\$300)")
plt.xlabel("Valores monetários $")
plt.ylabel("Quantidade de compras")
plt.show()
```

## como podemos colocar os valores maiores em exposição sem perder o estilo do gráfico

- log é sempre fazer a pergunta "10 elevado a quanto dá esse numero?
- log de base 1- = ng.log10


```python
log_purchases_values = np.log10(df_purchases["valor_compra"])

spots = [-1,0, 1, 2, 3]
names = ["$ O,10", "$ 1", "$ 10", "$ 100", "$ 1000"]

plt.hist(log_purchases_values, edgecolor='black', bins=50)
plt.xticks(spots, names)
plt.title("Histograma em log de valores de compras")
plt.xlabel("Valores $")
plt.ylabel("Quantidade de compras")
plt.show()
```

## Boxplot
